# Inventory EDA
This notebook uses the original Kaggle Retail Store Inventory Forecasting Dataset. It inspects shape, types, missing values, duplicates, distributions, correlations, product/store behavior, and trends.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv('../data/retail_store_inventory.csv', parse_dates=['Date'])
df.shape, df.isna().sum(), df.duplicated().sum()
```

```python
df.describe(include='all').T
```

```python
fig, ax = plt.subplots(1, 2, figsize=(13,4))
sns.histplot(df['Units Sold'], bins=40, ax=ax[0])
sns.scatterplot(data=df.sample(min(5000,len(df)), random_state=42), x='Units Sold', y='Inventory Level', hue='Category', alpha=.4, ax=ax[1])
plt.show()
```

```python
numeric=['Inventory Level','Units Sold','Units Ordered','Demand Forecast','Price','Discount','Competitor Pricing']
sns.heatmap(df[numeric].corr(), annot=True, cmap='vlag', center=0); plt.show()
```

```python
daily=df.groupby('Date').agg(sales=('Units Sold','sum'), inventory=('Inventory Level','mean'))
daily.plot(subplots=True, figsize=(12,6), title='Daily sales and mean inventory'); plt.show()
```
